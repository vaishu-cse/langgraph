Building AI Agent using Langgraph: using react agent.

previously we done this through the langraph. now it is moved to langchain.

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
# from langgraph.prebuilt import create_react_agent
from langchain.agents import create_agent
```

```python
@tool
def multiply(a: int, b:int)-> int:
    """this will multiply the given 2 numbers and return the number result"""
    print(f"[multiply tool] a: {a}, b: {b}")
    return a*b

@tool
def add(a:int, b:int) -> int:
    """it will add the given 2 numbers and returns the output number."""
    print(f"[add] a: {a}, b: {b}")
    return a+b

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")

tools = [add, multiply]

agent = create_agent(llm, tools)
```

```python
from langchain_core.messages import HumanMessage

result = agent.invoke({"messages": [HumanMessage(content="what is the value for 6 * 7 +10?")]})

print(f"final result from the agent is:: {result["messages"][-1].content}")
```
